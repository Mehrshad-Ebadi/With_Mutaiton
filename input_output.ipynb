```python
df = pd.read_csv('./Outputs/st_1000,nn_1000,mu_0.100000,ref_env_=0.000000/numrun_0/du_Out.txt', sep='\t', header=None)
```

```python
df = pd.DataFrame(df)
```

```python
df
```

```python
sn.violinplot(df,x=df[0],y=df[1])
plt.xlabel('Input')
plt.ylabel('Mean Output')
plt.yticks(np.arange(-1.3, 1.3, 0.1))
plt.grid(linewidth=0.2, axis='y')
plt.savefig('./in_out.jpeg', dpi=400)
```
