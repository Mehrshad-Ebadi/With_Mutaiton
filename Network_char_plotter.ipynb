```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib
```

```python
df = pd.read_csv('./Outputs/00Net_analysis.txt', sep=',')
dff = pd.read_csv('./Outputs/00_du_Net_analysis.txt', sep=',')
```

```python
df = pd.DataFrame(df)
dff = pd.DataFrame(dff)
```

```python
df
```

```python
import matplotlib as pl
ax = sns.barplot(x='step', y='cluster_coeffcient', data=df, label='single')
ax = sns.barplot(x='step', y='cluster_coeffcient', data=dff, label='double',alpha=0.4)
plt.xlabel('diameter')
every_nth = 10
for n, label in enumerate(ax.xaxis.get_ticklabels()):
    if n % every_nth != 0:
        label.set_visible(False)
plt.savefig('./diameter.jpeg')
```
